```python
%run add.ipynb

def sumlist(lst):
    total = 0
    for i in range(len(lst)):
        total = add(total, lst[i])
    return total

```

```python
list = [1, 2, 3, 4, 5]
result = sumlist(list)
print(result)
```

```python
list2 = [10, 31, 93, 4, 17]
result2 = sumlist(list2)
print(result2)
```
